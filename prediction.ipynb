```python
import re
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model

MAX_FEATURES = 10000
MAX_LEN = 500

word_index = imdb.get_word_index()
model = load_model('simple_rnn_imdb.h5')
```

```python
def preprocess_text(text):
    words = re.findall(r"[a-z0-9']+", text.lower())
    encoded = [1]  # start token
    for w in words:
        idx = word_index.get(w)
        if idx is None or idx + 3 >= MAX_FEATURES:
            encoded.append(2)  # OOV
        else:
            encoded.append(idx + 3)
    return sequence.pad_sequences([encoded], maxlen=MAX_LEN)

def predict_sentiment(review):
    prediction = model.predict(preprocess_text(review), verbose=0)
    score = float(prediction[0][0])
    sentiment = 'Positive' if score > 0.5 else 'Negative'
    return sentiment, score
```

```python
tests = [
    "This movie was fantastic! The acting was great and the plot was thrilling",
    "the movie was scary, i did not like it",
    "Terrible, boring and a waste of time",
]
for t in tests:
    s, sc = predict_sentiment(t)
    print(f'{s} ({sc:.4f}) -> {t}')
```
